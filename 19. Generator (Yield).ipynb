```python
# generator seperti iterable list,tuple, set , dict dll
# generator lebih efisien dalam menggunakan memory karna akan hilang ketika sudah di return
# cara panggil bisa di execute satu per satu
# akan hilang nilai yield apabila sudah di keluarkan 
```

```python
def func_gen():
    yield 1
    yield 2
    yield 3
    yield 4
```

```python
print(func_gen())
```

```python
for x in func_gen():
    print(x)
```

```python
#di execute 1 per 1 pake next()
vargen = func_gen()
```

```python
next(vargen)
```

```python
def func_gen(n):
    for i in range(n):
        yield i
```

```python
limasaja = func_gen(5)

for x in limasaja:
    print(x)
```

```python
print(list(limasaja)) #akan kosong karna sudah di keluarkan
print(tuple(limasaja)) #akan kosong karna sudah di keluarkan
```

```python
enamsaja = func_gen(6)
```

```python
print(list(enamsaja))
print(tuple(enamsaja)) 
```

```python
# filter kata, kata yg terfilter akan di yield

def find_kata(file, kata):
    with open(file, mode='r') as file:
        for x in file:
            if kata in x:
                yield x
```

```python
kata2 = find_kata('yieldExample.txt','Python')
```
